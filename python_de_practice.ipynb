```python
import psycopg2
```

```python
conn = psycopg2.connect(
    host="localhost",
    database="python-de-practice.",
    user="postgres",
    password="Root",
    port="5432"
)

print("Connected successfully!")
```

```python
# Cretae a cursor to execute SQL queries
cursor = conn.cursor()
```

```python
# Create customers table
cursor.execute("""
CREATE TABLE IF NOT EXISTS customers (
    customer_id VARCHAR(20),
    first_name VARCHAR(100),
    last_name VARCHAR(100),
    email VARCHAR(255),
    phone VARCHAR(20),
    city VARCHAR(100),
    state VARCHAR(50),
    signup_date DATE,
    segment VARCHAR(50),
    total_orders INT,
    total_spent NUMERIC(10,2),
    notes TEXT,
    signup_year INT,
    signup_month INT,
    signup_day INT
)
""")

# Save changes
conn.commit()

print("Customers table created successfully!")
```

```python
import pandas as pd

df = pd.read_csv("cleaned_customers.csv")

print(df.head())
print(df.shape)
```

```python
for row in df.itertuples(index=False, name=None):
    cursor.execute("""
        INSERT INTO customers (
            customer_id,
            first_name,
            last_name,
            email,
            phone,
            city,
            state,
            signup_date,
            segment,
            total_orders,
            total_spent,
            notes,
            signup_year,
            signup_month,
            signup_day
        )
        VALUES (%s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s)
    """, row)

conn.commit()

print("All 1000 records inserted successfully!")
```

```python
cursor.execute("SELECT COUNT(*) FROM customers")

result = cursor.fetchone()

print("Total records:", result[0])
```

```python
cursor.execute("SELECT * FROM customers LIMIT 5")

records = cursor.fetchall()

for record in records:
    print(record)
```

```python
# Update customer city
cursor.execute("""
UPDATE customers
SET city = 'Chennai'
WHERE customer_id = 'CU10002'
""")

conn.commit()

print("Customer updated successfully!")
```
