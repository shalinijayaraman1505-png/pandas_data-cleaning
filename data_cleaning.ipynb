```python

import pandas as pd

# Load dataset
df = pd.read_csv("messy-customers-1000.csv")


# 1. Inspect data
def inspect_data(df):
    print("First 5 Records")
    print(df.head())
    return df


# 2. Check dataset shape
def check_shape(df):
    print("\nDataset Shape")
    print(df.shape)
    return df


# 3. Check dataset information
def check_info(df):
    print("\nDataset Information")
    df.info()
    return df


# 4. Check missing values
def check_missing(df):
    print("\nMissing Values")
    print(df.isnull().sum())
    return df


# 5. Check duplicate rows
def check_duplicates(df):
    print("\nDuplicate Rows")
    print(df.duplicated().sum())
    return df


# 6. Remove exact duplicates
def remove_duplicates(df):
    df = df.drop_duplicates(keep="first")
    print("Exact duplicate rows removed.")
    return df


# 7. Clean first and last names
def clean_names(df):
    df["first_name"] = df["first_name"].str.strip().str.title()
    df["last_name"] = df["last_name"].str.strip().str.title()
    return df


# 8. Clean email
def clean_email(df):
    df["email"] = df["email"].str.strip().str.lower()
    return df


# 9. Clean phone numbers
def clean_phone(df):
    df["phone"] = df["phone"].str.replace("(", "", regex=False)
    df["phone"] = df["phone"].str.replace(")", "", regex=False)
    df["phone"] = df["phone"].str.replace("-", "", regex=False)
    df["phone"] = df["phone"].str.replace(" ", "", regex=False)
    return df


# 10. Clean city
def clean_city(df):
    df["city"] = df["city"].str.strip().str.title()
    return df


# 11. Clean state
def clean_state(df):
    state_mapping = {
        "California": "CA",
        "Texas": "TX",
        "New York": "NY",
        "Florida": "FL",
        "Pennsylvania": "PA",
        "New Jersey": "NJ",
        "Oregon": "OR",
        "Ohio": "OH",
        "Nevada": "NV",
        "Arizona": "AZ",
        "Utah": "UT",
        "Illinois": "IL",
        "Indiana": "IN",
        "Louisiana": "LA",
        "Georgia": "GA",
        "Missouri": "MO",
        "Kentucky": "KY",
        "Nebraska": "NE",
        "Wisconsin": "WI",
        "Massachusetts": "MA",
        "Rhode Island": "RI",
        "Colorado": "CO",
        "Virginia": "VA",
        "Connecticut": "CT",
        "Minnesota": "MN",
        "Michigan": "MI"
    }

    df["state"] = df["state"].replace(state_mapping)
    return df


# 12. Clean customer segment
def clean_segment(df):
    df["segment"] = df["segment"].str.strip().str.title()
    return df


# 13. Clean signup date
def clean_dates(df):
    df["signup_date"] = pd.to_datetime(
        df["signup_date"],
        format="mixed",
        errors="coerce"
    )
    return df


# 14. Extract date parts
def extract_date_parts(df):
    df["signup_year"] = df["signup_date"].dt.year
    df["signup_month"] = df["signup_date"].dt.month
    df["signup_day"] = df["signup_date"].dt.day
    return df


# 15. Clean total spent
def clean_total_spent(df):
    df["total_spent"] = df["total_spent"].str.replace(
        "$", "", regex=False
    )

    df["total_spent"] = pd.to_numeric(
        df["total_spent"],
        errors="coerce"
    )
    return df


# 16. Clean total orders
def clean_total_orders(df):
    df["total_orders"] = pd.to_numeric(
        df["total_orders"],
        errors="coerce"
    )
    return df


# 17. Fill missing numeric values
def fill_missing_values(df):
    spent_median = df["total_spent"].median()
    orders_median = df["total_orders"].median()

    df["total_spent"] = df["total_spent"].fillna(spent_median)
    df["total_orders"] = df["total_orders"].fillna(orders_median)

    df["total_orders"] = df["total_orders"].astype(int)

    return df


# 18. Handle missing email and notes
def handle_email_notes(df):
    df["email"] = df["email"].fillna("Not Provided")
    df["notes"] = df["notes"].fillna("Not Provided")
    return df


# Call all functions
inspect_data(df)
check_shape(df)
check_info(df)
check_missing(df)
check_duplicates(df)

df = remove_duplicates(df)
df = clean_names(df)
df = clean_email(df)
df = clean_phone(df)
df = clean_city(df)
df = clean_state(df)
df = clean_segment(df)
df = clean_dates(df)
df = extract_date_parts(df)
df = clean_total_spent(df)
df = clean_total_orders(df)
df = fill_missing_values(df)
df = handle_email_notes(df)


# Final data validation
print("\nFinal Dataset")
print(df.head())

print("\nFinal Dataset Shape")
print(df.shape)

print("\nFinal Data Types")
print(df.dtypes)

print("\nFinal Missing Values")
print(df.isnull().sum())

print("\nRemaining Exact Duplicates")
print(df.duplicated().sum())


# Save cleaned data into a new CSV
df.to_csv("cleaned_customers.csv", index=False)

print("\nCleaned dataset saved successfully!")
```
